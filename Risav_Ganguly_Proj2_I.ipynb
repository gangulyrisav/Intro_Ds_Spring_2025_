**SECTION II: DATA CLEANING & INDEXING (10 POINTS)**

```python
# 4. Remove duplicates
print("Number of duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
# Drop clearly redundant columns
columns_to_drop = ["Time period", "Observation value", "Revenue code", "Decimals",
    "Currency", "STRUCTURE", "Observation status", "STRUCTURE_ID", "STRUCTURE_NAME", "Unit multiplier", "UNIT_MULT", "FREQ", "OBS_STATUS", "REVENUE_CODE", "DECIMALS", "CURRENCY", "UNIT_MEASURE", "CTRY_SPECIFIC_REVENUE", "ACTION", "STANDARD_REVENUE", "Country-specific revenue category", "MEASURE"] 
df = df.drop(columns=columns_to_drop)
df_new = df.copy()
df_new
```

```python
# 5. Two examples of filtering using logical conditions
# Filter for a specific group of countries (e.g., OECD)
oecd_countries = ['Germany', 'France', 'Japan', 'Canada', 'United States', 'United Kingdom']
df_oecd = df_new[df_new["Reference area"].isin(oecd_countries)]
df_oecd.head()
```

```python
# Filter for data from the year 2020 onward
df_recent = df_new[df_new["TIME_PERIOD"] >= 2020]
df_recent.head()
```

**SECTION III: DESCRIPTIVE STATISTICS (15 POINTS)**

```python
# Question 6 
numerical_stats = df_new['OBS_VALUE'].describe()
print(numerical_stats)
```

**Interesting Insight:**
The maximum tax revenue as a percentage of GDP (45.97% in Denmark, 2013) is significantly higher than the minimum (1.198% in Somalia, 2016), highlighting a vast disparity in tax collection efficiency or fiscal policy across countries. This suggests that developed nations tend to have much higher tax-to-GDP ratios compared to less developed ones.

```python
# 8. Groupby analysis

# Mean revenue by country
df_new.groupby("Reference area")["OBS_VALUE"].mean().sort_values(ascending=False).head()

# Count of records by revenue category
df_new.groupby("Revenue category")["OBS_VALUE"].count().sort_values(ascending=False).head()
```

```python
# Count occurrences of each country
country_counts = df_new['REF_AREA'].value_counts()
print(country_counts.head())

# Count occurrences of each year
year_counts = df_new['TIME_PERIOD'].value_counts()
print(year_counts)
```

**Observation:**
The TIME_PERIOD column shows a almost common frequency of data collection for 2013 (one less) and later years in the sample. This confirms the uniformity in the data.

```python
# Additional Grouping: (Africa" vs. individual countries) to compare averages.
# Filter for Africa and OECD average for comparison
region_group = df[df['REF_AREA'].isin(['F', 'OECD_REP'])].groupby('REF_AREA')['OBS_VALUE'].mean()
print(region_group)
```

**Insight:**    
OECD countries have a significantly higher average tax-to-GDP ratio (33.5%) compared to the African average (~15.3%), reflecting differences in economic development and tax systems.

```python
# Checking the most volatitle nations
volatility = df.groupby('REF_AREA')['OBS_VALUE'].std()
print(volatility.sort_values(ascending=False).head())
```

```python
# Checking the least volatile nations
volatility = df.groupby('REF_AREA')['OBS_VALUE'].std()
print(volatility.sort_values(ascending=False).tail())
```

```python
stable_countries = df.groupby('REF_AREA')['OBS_VALUE'].std().sort_values().head(5)
print("Most Stable:\n", stable_countries)
print("Most Volatile:\n", volatility.sort_values(ascending=False).head())
```
